# Chapter 3. Interpolation and Polynomial Approximation
## 3.1 Interpolation and the Lagrange Polynomial

If $x_0,\ x_1,\ \cdots,\ x_n$ are $n+1$ distinct numbers and $f$ is a function whose values are given at these numbers, then a unique polynomial $P(x)$ of degree at most $n$ exists with
$$f(x_k) = P(x_k),\qquad \text{for each}\ k = 0,1,\cdots,n.$$	
This polynomial is given by
$$P(x) = f(x_0) \ell_{0}(x) + \cdots + f(x_n) \ell_{n}(x) = \sum_{k=0}^n f(x_k) \ell_{k}(x),$$
where, for each $k=0,1,\cdots,n$,
$$\ell_{k}(x) = \frac{(x-x_0)(x-x_1)\cdots(x-x_{k-1})(x-x_{k+1})\cdots(x-x_n)}{(x_k-x_0)(x_k-x_1)\cdots(x_k-x_{k-1})(x_k-x_{k+1})\cdots(x_k-x_n)} = \prod_{\substack{i=0\\ i\neq k}}^n \frac{(x-x_i)}{(x_k - x_i)}.$$

````python
def Lagrange_polynomial(t,x,fval):
    p = 0
    for i in range(len(x)):
        l_i = 1
        for l in range(len(x)):
            if i == l:
                continue
            l_i *= ((t - x[l]) / (x[i] - x[l]))
        p += fval[i] * l_i
    return p
````

### Example
The polynomial of degree $\le 2$ that passes through the three points $(0,1)$, $(-1,2)$, and $(1,3)$ is

````python
x = np.array([0, -1, 1])
fval = np.array([1,2,3])
t = 0.5
val = Lagrange_polynomial(t,x,fval)
print(val)
````

````python
t = np.linspace(-1.2,1.2,200)
val = np.zeros_like(t)
for j in range(len(t)):
    val[j] = Lagrange_polynomial(t[j],x,fval)
````

````python
plt.figure(figsize=(12,10))
plt.plot(t,val, label = 'Lagrange Interpolating Polynomial')
plt.plot(x,fval,'o')
plt.legend()
plt.title('Lagrange Interpolating Polynomial with $x_0$ = {}, $x_1$ = {}, and $x_2$ = {}'.format(x[0],x[1],x[2]))
````

## 3.3 Divided Differences
### Newton's Divided Difference Formula

Divdif ($d$, $x$, $n$)<br>
Remark: $d$ and $x$ are vectors with entries $f(x_i)$ and $x_i$, $i = 0, 1, \dots, n$, respectively. On exit, $d_i$ will contain $f[x_0,\dots, x_i]$.<br>
1. Do through Step 4 for $i = 1, 2, \dots, n$
2. Do through Step 4 for $j = n, n-1, \dots, i$
3. $d_j = (d_j - d_{j-1}) / (x_j - x_{j-i})$
4. Exit from the algorithm

````python
def Divdif(fval,x,n):
    result = fval.copy()
    for i in range(1,n+1):
        for l in reversed(range(i,n+1)):
            result[l] = (result[l] - result[l-1]) / (x[l] - x[l-i])
    return result
````

Interp ($d$, $x$, $n$, $t$, $p$)<br>
Remark: On entrance, $d$ and $x$ are vectors containing $f[x_0,\dots, x_i]$. and $x_i$, $i = 0, 1, \dots, n$, respectively. On exit, $p$ will contain the value $p_n(t)$ of the $n$th-degree polynomial interpolating $f$ on $x$.
1. $p = d_n$
2. Do through Step 4 for $i = n-1, n-2, \dots, 0$
3. $p = d_i + (t-x_i) p$
4. Exit the algorithm

````python
def interp(d,x,n,t):
    p = d[-1]
    for i in reversed(range(n)):
        p = d[i] + (t-x[i]) * p
    return p
````

### Example

Construct a divided difference table for $f(x) = \sqrt{x}$.
Then find approximations with $n = 1, 2, 3,$ and $n = 4$ at $x = 2.05, 2.15$, and $x=3$.

## 3.4 Hermite Interpolation

Let
\begin{align*}
\tilde{h}_i(x) &= (x-x_i)[\ell_i(x)]^2 \\[2ex]
h_i(x) &= (1 - 2\ell'_i(x_i)(x-x_i))[\ell_i(x)]^2
\end{align*} 
Then
$$H_n(x) = \sum_{i=1}^n y_i h_i(x) + \sum_{i=1}^n y'_i \tilde{h}_i(x)$$
<br>

Newton divided difference form
\begin{align*}
H_n(x) &= f(x_1) + (x-x_1)f[x_1, x_1] + (x-x_1)^2 f[x_1, x_1, x_2] + \cdots \\
&\qquad + (x-x_1)^2 \cdots (x-x_{n-1})^2 (x-x_n) f[x_1, x_1, x_2, x_2, \cdots, x_n, x_n]
\end{align*}

````python
def Divdif_Hermite(fval,dfval,x):
    n = len(x)

    z = np.zeros(n*2)
    d = np.zeros(n*2)
    for i in range(n):
        z[i*2] = x[i]
        z[i*2+1] = x[i]

        d[i*2] = fval[i]
        d[i*2+1] = fval[i]
    
    for i in range(1, n*2):
        for l in reversed(range(i,n*2)):
            if z[l] == z[l-i]:
                d[l] = dfval[(l-1)//2]
            else:
                d[l] = (d[l] - d[l-1]) / (z[l] - z[l-i])
                
    return d, z
````

## 3.5 Cubic Spline Interpolation

### Complete Cubic Spline (Clamped Cubic Spline)

$$S_i(x) = a_i + b_i(x-x_i) + c_i(x-x_i)^2 + d_i(x-x_i)^3$$

````python
def clamped_cubic_spline(n,x,fval,FPO,FPN):

    h = np.zeros(n)
    M = np.zeros(n+1)
    D = np.zeros(n+1)
    A = np.zeros((n+1,n+1))
    y = np.zeros(n+1)

    y[0] = fval[0]
    for i in range(n):
        h[i] = x[i+1] - x[i]
        y[i+1] = fval[i+1]
            
        if i == 0:
            D[i] = (y[i+1] - y[i]) / h[i] - FPO
            A[i][i] = h[i] / 3
            A[i][i+1] = h[i] / 6
        else:
            D[i] = (y[i+1] - y[i]) / h[i] - (y[i] - y[i-1]) / h[i-1] 
            A[i][i-1] = h[i-1] / 6
            A[i][i] = (h[i-1] + h[i]) / 3
            A[i][i+1] = h[i] / 6

    D[n] = FPN - (y[n] - y[n-1]) / h[n-1]
    A[n][n-1] = h[n-1] / 6
    A[n][n] = h[n-1] / 3
    
    M = np.dot(np.linalg.inv(A), D)

    a = np.zeros(n+1)
    b = np.zeros(n)
    c = np.zeros(n+1)
    d = np.zeros(n)

    for i in range(n):
        a[i] = y[i]
        b[i] = (y[i+1] - y[i]) / h[i] - h[i] * (2 * M[i] + M[i+1]) / 6
        c[i] = M[i] / 2
        d[i] = (M[i+1] - M[i]) / (6 * h[i])
    a[n] = y[n]
    c[n] = M[n] / 2
    return a, b, c, d
````

````python
def evaluate_cubic_spline(nodes,x,a,b,c,d):
    val = np.zeros_like(nodes)
    for i in range(len(x)-1):
        ind = (nodes >= x[i]) & (nodes <= x[i+1])
        xx = nodes[ind]
        val[ind] = a[i] + b[i]*(xx - x[i]) + c[i]*(xx-x[i])**2 + d[i]*(xx-x[i])**3
    return val
````

### Example 4
Example 2 used a natural spline and the data points $(0,1),\ (1,e),\ (2,e^2),$ and $(3,e^3)$ to form a new approximating function $S(x)$. Determine the clamped spline $s(x)$ that uses this data and the additional information that, since $f'(x) = e^x$, so $f'(x) = 1$ and $f'(e) = e^3$.

````python
n = 3
x = np.array([0,1,2,3])
fval = np.exp(x)
FPO = 1
FPN = np.exp(3)
````

````python
a,b,c,d = clamped_cubic_spline(n,x,fval,FPO,FPN)
print('a = \n', a)
print('b = \n', b)
print('c = \n', c)
print('d = \n', d)
````
